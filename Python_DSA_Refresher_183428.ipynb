#1. Write a program that asks the user for a number of days. The program then prints out the number of seconds in the amount of days given.

```python
#Converting number of seconds to amt of days
# d = days; s = seconds;
d = int(input("Please enter the number of days:"))
s = d * 24 * 60 * 60
print(f"{d} day(s) = {s} seconds")
```

#2. Write a program that asks a user to input the radius then the program calculates the volume of a sphere(the formula for the volume is (4/3)πr3). Use the exponential operator in python to compute (r3).

```python
import math
# volume of a sphere is (4/3)pi*r^3
# r = Radius; v = Volume;

r = float(input("please enter the radius of the sphere: "))
v = (4/3) * math.pi * r ** 3  #Exponential operator **
print(f"The final answe hence the volume of a sphere with radius {r} is {v:.2f}")
```

# 3. Using functions, write a program to compute the area and perimeter of a square. The program should ask the user to enter a number corresponding to the side length of the square and display the area and perimeter of the square.

```python
def sqr_area(s):
  """The area of a square: s times s."""
  return s * s

def sqr_perimeter(s):
    """Return the perimeter of a square: four times the side(s)."""
    return 4 * s

s = float(input("please enter the side length of the square: "))
print(f"Area is: {sqr_area(s)}")
print(f"Perimeter is: {sqr_perimeter(s)}")
```

# 4. Write a program using functions that determines whether a character input by a user is uppercase or lower case.

```python
def check_the_case(ch):
  """Check case if in Lowercase, Uppercase or none"""
  if ch.isupper():
    return "Uppercase"
  elif ch.islower():
        return "lowercase"
  else:
        return "it is not an alphabetic character; Try Again"

ch = input("Enter a single character: ")


if len(ch) != 1:
    print("Please enter exactly one character.")
else:
    print(f"'{ch}' is {check_the_case(ch)}")
```

#6. Write a Python program that does the following:
##• Uses a loop for a user to continually input 5 values to populate an array.
##• Calculates and displays the average of the values input into the array.

```python
v = []
for i in range(5):

    val = float(input(f"Please enter the value {i + 1}: "))
    v.append(val)

average = sum(v) / len(v)

print(f"Values entered: {v}")
print(f"Average = {average:.2f}")
```
